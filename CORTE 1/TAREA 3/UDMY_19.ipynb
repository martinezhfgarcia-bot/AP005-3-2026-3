```python
import pandas as pd
insta = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/Instagram.csv', parse_dates=["reg date", "last login"])
insta["gender"] = insta["gender"].astype("category")
insta["Country"] = insta["Country"].astype("category")
insta["Influencer"] = insta["Influencer"].astype("bool")
insta.head()
```

* *import pandas as pd: importa pandas.*

* *nsta = pd.read_csv(...): lee el CSV, convierte reg date y last login a tipo fecha.*

* *insta["gender"] = ...: convierte gender a categoría.*

* *insta["Country"] = ...: convierte Country a categoría.*

* *insta["Influencer"] = ...: convierte Influencer a booleano.*

* *insta.head(): muestra las primeras filas.*

```python
mask = (insta["gender"] == "Male") & (insta["reg date"] > "2015-12-31") & (insta["Country"] == "United States")
insta[mask]
```

* *mask = ...: crea una máscara booleana con tres condiciones: género masculino, fecha de registro posterior a 2015-12-31 y país Estados Unidos.*

* *insta[mask]: filtra el DataFrame usando la máscara.*

```python
mask1 = insta["gender"] == "Male"
mask2 = insta["reg date"] > "2015-12-31"
mask3 = insta["Country"] == "United States"
insta[mask1 & mask2 & mask3]
```

* *mask1: condición género masculino.*

* *mask2: condición fecha de registro > 2015-12-31.*

* *mask3: condición país Estados Unidos.*

* *insta[mask1 & mask2 & mask3]: combina las tres máscaras con AND y filtra.*

```python
mask = (insta["gender"] == "male") & (insta["reg date"] > "2015-12-31")
insta[mask]
```

* *mask: usa "male" en minúscula (no coincide con "Male").*

* *insta[mask]: filtra, pero como no hay coincidencias, devuelve un DataFrame vacío.*
