# Creación Estructurada de DataFrames
*Este módulo explica cómo construir tablas de datos desde cero utilizando listas básicas de Python.*

```python
import pandas as pd
```

*Importa nuevamente la librería pandas para tener habilitadas sus funciones de creación de tablas.*

```python
numbers = [1, 2, 3, 4, 5, 6]
```

*Crea una lista tradicional de Python que contiene una serie sencilla de números enteros.*

```python
pd.DataFrame(numbers)
```

*Transforma la lista plana de números en un DataFrame (una tabla formal de una sola columna).
Pandas le asigna automáticamente un encabezado y números de fila empezando desde el cero.*

```python
data = [["Jhon",  23], ["Mary", 18], ["Jhon", 31]]
```

*Crea una variable llamada 'data' que guarda una lista de listas.
Cada sub-lista representa una fila independiente que contiene un nombre y una edad.*

```python
pd.DataFrame(data)
```

*Convierte esa matriz de listas en una tabla de datos de dos dimensiones.
Por defecto, los nombres de las columnas serán asignados con los números 0 y 1.*

```python
pd.DataFrame(data, columns = ["name", "age"])
```

*Vuelve a crear la tabla a partir de los datos, pero ahora usa el atributo 'columns'.
Esto nos permite nombrar la primera columna como "name" y la segunda como "age" para que sea legible.*

```python
pd.DataFrame(data, columns = ["name", "age"], index = ["id1", "id2", "id3"])
```
