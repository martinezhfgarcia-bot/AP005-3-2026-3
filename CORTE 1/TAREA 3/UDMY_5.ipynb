# Selección de Múltiples Columnas en Pandas
*Este documento explica cómo filtrar una tabla grande para visualizar únicamente las columnas que nos interesan analizar.*

```python
import pandas as pd
```

*Importa la librería pandas al inicio para poder usar sus funciones.*

```python
nba = pd.read_excel("Nba Teams.xlsx")
```

*Carga el archivo Excel con los datos de los jugadores de la NBA y lo almacena en la variable 'nba'.*

```python
nba.head()
```

*Muestra en pantalla las primeras cinco filas de la tabla original, la cual contiene todas las columnas disponibles.*

```python
nba[["Name", "Team"]]
```

*Selecciona y muestra únicamente dos columnas específicas: "Name" (Nombre) y "Team" (Equipo).
Se debe utilizar doble corchete [[ ]] porque estamos pasando una lista de columnas a buscar dentro del DataFrame.*

```python
nba[["Name", "Team", "Salary"]]
```

*Realiza una selección múltiple más amplia, extrayendo los nombres, equipos y salarios de todos los jugadores.
El doble corchete es fundamental siempre que se quiera aislar más de una sola columna a la vez.*
