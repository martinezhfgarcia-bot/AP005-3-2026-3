### Selección y operaciones matemáticas básicas


```python
nba.fillna(0)
```

*Reemplaza absolutamente todos los valores vacíos (NaN) que existan en cualquier columna del DataFrame con el número 0. Al no usar inplace=True, esto solo genera una vista previa y no modifica la variable original.*

```python
nba[["College"]].fillna(0, inplace = True)
```

* *nba[["College"]]: Delimita la acción únicamente a la columna "College".*

* *.fillna(0, inplace = True): Rellena los datos faltantes de esa columna con un 0. El parámetro inplace=True obliga a Pandas a guardar este cambio definitivamente en la memoria del DataFrame original.*

```python
nba["Name"].fillna("Missing name", inplace = True)
```

* *nba["Name"].fillna(...): Selecciona la columna "Name" y busca celdas vacías.*

* *("Missing name", inplace = True): Sustituye los valores nulos encontrados por la cadena de texto "Missing name", sobreescribiendo los datos permanentemente en la tabla original debido a la instrucción inplace=True.*
