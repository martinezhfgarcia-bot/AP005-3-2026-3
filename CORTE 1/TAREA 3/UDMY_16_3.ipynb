*Extrae y devuelve un arreglo con los valores únicos contenidos en la columna Gender ('Male' y 'Female').*

```python
df.pivot_table(values = "Amount", index = "Gender", aggfunc="sum")
df.pivot_table(values = "Amount", index = "Gender", aggfunc="mean")
```

* *df.pivot_table(...): Crea una tabla dinámica agrupando la información.*

* *values = "Amount": Define la columna numérica sobre la cual se realizarán los cálculos estadísticos.*

* *index = "Gender": Establece la columna Gender como categoría de agrupación para las filas.*

* *aggfunc="sum" / aggfunc="mean": Aplica funciones de agregación para obtener la suma total y el promedio de los montos por cada género*

```python
df.pivot_table(values = "Amount", index = "Menu", aggfunc="sum")
```

*Genera una tabla dinámica que calcula la suma total de los montos (Amount) agrupados específicamente por cada tipo de platillo o producto del menú (Menu).*

```python
df.pivot_table(values = "Amount", index = ["Gender", "Menu"], aggfunc="sum")
```

*Implementa una tabla dinámica con agrupación multinivel en las filas, utilizando tanto Gender como Menu, para sumar los montos correspondientes a cada combinación.*

```python
df.pivot_table(values = "Amount", index = ["Gender", "Menu"], columns = "City", aggfunc="sum")
```

*Extiende la tabla dinámica anterior distribuyendo los datos en columnas independientes según la ciudad, permitiendo cruzar género, menú y ubicación geográfica.*

```python
df.pivot_table(values = "Amount", index = ["Gender", "Menu"], columns = ["Frequency", "City"], aggfunc="sum")
```
