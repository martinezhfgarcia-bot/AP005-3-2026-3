*Combinación de DataFrames con merge (clientes, ventas y productos).*

```python
import pandas as pd
customers = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/ABC - Customers.csv')
sales = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/ABC - Sales.csv')
```

* *import pandas as pd: importa pandas con el alias pd.*

* *customers = pd.read_csv(...): lee el CSV de clientes y lo guarda en customers.*

* *sales = pd.read_csv(...): lee el CSV de ventas y lo guarda en sales.*

```python
customers
sales
```

* *customers: muestra el DataFrame de clientes.*

* *sales: muestra el DataFrame de ventas.*

```python
cust_sales = pd.merge(customers, sales, how = 'inner', on = 'cust_id')
cust_sales
```

* *cust_sales = pd.merge(customers, sales, how = 'inner', on = 'cust_id'): hace un inner join entre customers y sales usando cust_id.*

* *cust_sales: muestra el resultado del merge.*

```python
#inner join
cust_sales = pd.merge(customers, sales[['cust_id', 'product', 'qty', 'store_location']], how = 'inner', on = 'cust_id')
cust_sales
```

* *#inner join: comentario indicando que se hace un inner join.*

* *cust_sales = pd.merge(customers, sales[['cust_id', 'product', 'qty', 'store_location']], how = 'inner', on = 'cust_id'): otro inner join, pero seleccionando solo algunas columnas de sales.*

* *cust_sales: muestra el resultado.*

```python
#left join
customers['cust_id'].nunique()
sales['cust_id'].nunique()
```

* *#left join: comentario indicando que sigue un left join.*

* *customers['cust_id'].nunique(): cuenta cuántos cust_id únicos hay en customers.*

* *sales['cust_id'].nunique(): cuenta cuántos cust_id únicos hay en sales.*

```python
df1 = pd.merge(customers, sales, how = 'left', on = 'cust_id')
df1
```

* *df1 = pd.merge(customers, sales, how = 'left', on = 'cust_id'): hace un left join: conserva todos los clientes, aunque no tengan ventas.*

* *df1: muestra el resultado.*

```python
df1['sale_id'].isnull()
df1[df1['sale_id'].isnull()]['customer']
```

* *df1['sale_id'].isnull(): devuelve una Serie booleana indicando qué ventas son nulas.*

* *df1[df1['sale_id'].isnull()]['customer']: muestra los clientes que no tienen ventas (sale_id nulo).*

```python
pd.merge(customers, sales, how = "right", on = 'cust_id')
pd.merge(customers, sales, how = "outer", on = 'cust_id')
```

* *pd.merge(customers, sales, how = "right", on = 'cust_id'): hace un right join: conserva todas las ventas, aunque no tengan cliente.*

* *pd.merge(customers, sales, how = "outer", on = 'cust_id'): hace un outer join: conserva todos los registros de ambas tablas.*

```python
products = pd.read_csv('/content/drive/MyDrive/udmy/dataAnalisys/ABC - Products.csv')
products
```

* *products = pd.read_csv(...): lee el CSV de productos y lo guarda en products.*

* *products: muestra el DataFrame de productos.*

```python
pd.merge(customers, sales, how = 'inner', on = 'cust_id')
customers.merge(sales, on = 'cust_id').head(2)
```

* *pd.merge(customers, sales, how = 'inner', on = 'cust_id'): vuelve a hacer inner join entre customers y sales.*

* *customers.merge(sales, on = 'cust_id').head(2): usa el método .merge() de DataFrame y muestra las primeras 2 filas.*


```python
customers.merge(sales, on = 'cust_id').merge(products, on = 'prod_id').head(2)
customers.merge(sales, on = 'cust_id').merge(products[['prod_id', 'price']], on = 'prod_id').head(2)
```

* *Primera línea: encadena merges: customers + sales + products, y muestra 2 filas.*

* *Segunda línea: similar al anterior pero solo trae prod_id y price de productos.*
