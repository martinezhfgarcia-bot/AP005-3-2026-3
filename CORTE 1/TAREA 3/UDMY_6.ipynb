# Extracción de una Única Columna (Series)
*En este cuaderno aprenderemos las dos formas de extraer una sola columna de un DataFrame, la cual pandas maneja internamente como un objeto llamado 'Series'.*

```python
import pandas as pd
```

*Importa la librería estándar de análisis de datos.*

```python
nba = pd.read_excel("Nba Teams.xlsx")
```

*Lee el archivo Excel de la NBA y lo guarda en la memoria bajo el nombre de 'nba'.*

```python
nba.head()
```

*Imprime las primeras cinco filas completas para inspeccionar los nombres exactos de los encabezados de las columnas.*

```python
nba["Name"]
```

*Primera forma de selección (Notación por corchetes): extrae únicamente la columna con el nombre de los jugadores.
Como es una sola columna, se usa un solo par de corchetes.
El resultado visualmente ya no es una tabla bonita, sino una lista simple conocida como 'Series'.*

```python
nba["Salary"]
```

*Utiliza la notación por corchetes para extraer y mostrar solo la columna que contiene los salarios de la liga.*

```python
nba.Name
```

*Segunda forma de selección (Notación por punto): extrae la columna de nombres escribiendo un punto seguido del nombre de la columna.
Esta forma es más rápida de escribir, pero solo funciona si el nombre de la columna no tiene espacios (por ejemplo, fallaría si la columna se llamara "First Name").*
