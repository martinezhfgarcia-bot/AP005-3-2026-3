# Eliminación de Columnas en un DataFrame
*En este cuaderno aprenderemos cómo borrar de forma temporal o permanente columnas enteras usando la función '.drop()'.*

```python
import pandas as pd
```

*Importamos la librería pandas, esencial para cualquier tipo de manipulación de tablas.*

```python
nba = pd.read_excel("Nba Teams.xlsx")
```

*Cargamos el archivo original de la NBA y lo almacenamos en la variable principal 'nba'.*

```python
nba.head()
```

*Verificamos las primeras líneas para comprobar que la tabla cargó con todas sus columnas originales.*

```python
nba.drop("Salary", axis = 1)
```

*Utilizamos la función '.drop()' para decirle a pandas que queremos deshacernos de los datos.
Escribimos "Salary" para indicar qué queremos borrar, y 'axis = 1' le aclara al programa que estamos borrando una columna entera (si fuera 'axis = 0' borraría una fila).
Esta eliminación es temporal y solo afecta la visualización inmediata de esta celda.*

```python
nba
```

*Al imprimir nuevamente la tabla 'nba', comprobamos que la columna de Salarios sigue existiendo.
Esto confirma que el borrado anterior no fue definitivo.*

```python
nba.drop("Salary", axis = 1, inplace = True)
```

*Volvemos a usar '.drop()' para borrar el salario, pero esta vez agregamos el parámetro vital: 'inplace = True'.
Este parámetro fuerza a pandas a sobrescribir la tabla en la memoria, haciendo que el borrado sea definitivo.*

```python
nba
```

*Verificamos el cambio y comprobamos que la columna "Salary" ha desaparecido permanentemente de nuestro análisis.*

```python
nba.drop(["Team", "Division"], axis = 1, inplace = True)
```

*Utiliza '.drop()' para borrar múltiples columnas al mismo tiempo.
Para lograrlo, se debe pasar una lista (usando corchetes) con los nombres de las columnas que queremos eliminar definitivamente ('inplace = True').*

```python
nba
```

*Mostramos la tabla final, ahora mucho más limpia y compacta tras las eliminaciones.*
