# Creación de DataFrames usando Diccionarios
*Este bloque enseña cómo transformar un diccionario tradicional de Python en una tabla bidimensional de pandas.*

```python
import pandas as pd
```

*Importa la librería pandas al inicio del cuaderno para habilitar sus herramientas de datos.*

```python
data = {"name": ["Sally", "Jhon", "Kim"], "age": [4, 6 ,7], "gender": ["female", "male", "male"]}
```

*Crea un diccionario de Python y lo guarda en la variable 'data'.
Las claves del diccionario ("name", "age", "gender") se convertirán en los títulos de las columnas.
Los valores son listas que contienen la información respectiva hacia abajo para cada columna.*

```python
pd.DataFrame(data)
```

*Convierte el diccionario estructurado en un DataFrame real de pandas.
Al imprimirlo, las filas reciben automáticamente un índice numérico empezando desde el cero.*

```python
pd.DataFrame(data, index = ["Kid1", "Kid2", "Kid3"])
```
