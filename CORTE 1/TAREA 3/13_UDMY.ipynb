*(Ordenamiento de Datos)
Ordenamiento simple y visualización*

```python
nba.sort_values("Name").head(20)
```

* *nba.sort_values("Name"): Organiza todas las filas del DataFrame alfabéticamente basándose en los datos de la columna "Name".*

* *.head(20): Muestra únicamente los primeros 20 registros resultantes de ese ordenamiento para dar un vistazo rápido a la lista organizada.*

```python
nba.sort_values("Salary")
#nba.sort_values("Salary", ascending = False, inplace = True)
```

* *nba.sort_values("Salary"): Ordena la tabla tomando como referencia la columna "Salary", organizando los valores de menor a mayor por defecto.*

* *El código comentado (ascending = False, inplace = True) muestra cómo invertir el orden (de mayor a menor salario) y cómo utilizar el parámetro inplace para sobreescribir y guardar los cambios permanentemente en el DataFrame original.*

```python
nba.sort_values(["Team", "Name"])
#nba.sort_values(["Team", "Name"], na_position = 'first')
```

* *nba.sort_values(["Team", "Name"]): Realiza un ordenamiento con múltiples niveles de prioridad pasándole una lista de columnas. Primero organiza toda la tabla por la columna "Team" y, posteriormente, dentro de cada equipo, ordena a los jugadores alfabéticamente por su "Name".*

* *El parámetro comentado na_position = 'first' ilustra cómo forzar a Pandas para que coloque cualquier fila con valores nulos (NaN) al principio de la tabla, en lugar de su comportamiento predeterminado que es enviarlos al final.*
