Eliminación básica de valores nulos

```python
nbaNoNull = nba.dropna()
```

* *nba.dropna(): Identifica y elimina cualquier fila que contenga al menos un valor vacío (NaN) en cualquiera de sus columnas.*

* *nbaNoNull = ...: Guarda el resultado de esta limpieza exhaustiva en una nueva variable, manteniendo el DataFrame nba original intacto.*

```python
nba.dropna(how="all")
```

*Aplica un borrado menos estricto. Borra una fila únicamente si todas y cada una de sus celdas están completamente vacías (NaN).*

```python
nba.dropna(axis=0)
```

*El parámetro axis=0 especifica que la eliminación opera a nivel de filas horizontales (este es el comportamiento predeterminado de la función).*

```python
nba.dropna(axis=1)
```

*El parámetro axis=1 cambia la lógica de barrido. En lugar de filas, elimina columnas completas (verticales) si encuentran al menos un dato vacío en ellas.*

```python
nba.dropna(subset=["Name", "Salary", "College"])
```

*Restringe la condición de borrado a variables específicas. Solo elimina la fila si encuentra un valor vacío dentro de las columnas listadas (Name, Salary, College), ignorando los valores nulos que puedan existir en el resto de la tabla.*
