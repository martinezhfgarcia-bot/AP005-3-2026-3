*Ordena el DataFrame enfocándose estrictamente en un nivel particular del índice (nivel 0 por tipo de fruta o nivel 1 por color).*

*Facilita la selección y filtrado de datos utilizando las etiquetas del MultiIndex, permitiendo consultar registros específicos como las manzanas o una combinación exacta (manzanas verdes).*

```python
df.loc[("Apples", "Green"), "price usd"]
```

*Extrae un valor puntual de la celda utilizando una tupla para especificar la combinación exacta del índice jerárquico y el nombre de la columna objetivo.*

```python
df.stack()
df.stack().to_frame()
df2 = df.stack().to_frame()
df2
```

* *df.stack(): Reorganiza las columnas del DataFrame convirtiéndolas en un nuevo nivel dentro del índice de las filas (formato apilado o "long format").*

* *.to_frame(): Convierte la serie resultante nuevamente en un DataFrame estructurado y lo asigna a la variable df2.*


```python
df2.unstack()
```

Realiza la operación inversa a stack, desapilando los niveles del índice para devolverlos a la estructura de columnas tabulares tradicionales.
