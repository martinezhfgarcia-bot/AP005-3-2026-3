```python
sector.sum()
sector["Employees"].sum()
sector["Employees"].mean()
```

* *sector.sum(): suma los valores numéricos por cada grupo (puede fallar si hay columnas no numéricas).*

* *sector["Employees"].sum(): suma la columna Employees por cada grupo.*

* *sector["Employees"].mean(): calcula el promedio de Employees por cada grupo.*

```python
sector.agg({"Revenues" : "mean", "Profits" : "mean", "Employees" : "sum"})
sector.agg({"Revenues" : ["mean", "max"], "Profits" : "mean", "Employees" : "sum"})
sector.agg({"Revenues" : ["mean", "max"], "Profits" : ["max", "min"], "Employees" : "sum"})
```

* *Primera línea: aplica media a Revenues y Profits, y suma a Employees.*

* *Segunda línea: agrega media y máximo a Revenues, media a Profits, suma a Employees.*

* *Tercera línea: agrega media y máximo a Revenues, máximo y mínimo a Profits, suma a Employees.*

```python
fortune = df
sectors = fortune.groupby('Sector')
sector.size()
```

* *fortune = df: asigna df a fortune.*

```python
df = pd.DataFrame(columns = fortune.columns)
df
```

* *df = pd.DataFrame(columns = fortune.columns): crea un DataFrame vacío con las mismas columnas que fortune.*

* *df: muestra el DataFrame vacío.*

```python
for sector, values in sectors:
  company_with_highest_profit = values.nlargest(1, "Profits")
  df = df.append(company_with_highest_profit)
```

* *for sector, values in sectors:: itera por cada grupo del agrupamiento sectors.*

* *company_with_highest_profit = values.nlargest(1, "Profits"): obtiene la compañía con mayor Profits dentro del grupo.*
