```python
nba = pd.read_excel("Nba Teams.xlsx")
nba.head()
```

*Carga el archivo original en la variable 'nba' y muestra las primeras filas para revisar su estado inicial.*

```python
nba["Teams"] = " "
```

*Crea una segunda columna vacía llamada "nba Teams" y luego imprime el inicio de la tabla.

*Esta celda sirve como laboratorio de pruebas sobre la notación por punto.

La única línea activa extrae y muestra la columna 'Teams' que acabamos de crear (que está vacía).*

```python
nba["Sports"] = "basket"
nba.head()
```

*Agrega otra columna llamada "Sports" al final de la tabla.

```python
nba.drop("Teams", axis = 1, inplace = True)
```
