*# SECCIÓN DE TUPLAS*

*Las tuplas son idénticas a las listas, pero son inmutables (nadie puede modificarlas, agregarles ni borrarles elementos después de creadas).*

```python
print("###########################")
print("############TUPLAS#########")
my_tupla = tuple(my_lista)
print("my_tuple: ", my_tupla)
```

*Líneas 1 y 2: Imprimen líneas separadoras decorativas en la consola.*

*Línea 3: Usa la función tuple() para tomar nuestra lista de colores y transformarla permanentemente en una tupla, guardándola en la variable 'my_tupla'.*

*Línea 4: Imprime la nueva tupla. En pantalla se notará la diferencia porque estará encerrada entre paréntesis () en lugar de corchetes [].*


```python
print(my_tupla[0])
print(my_tupla[2])
```

*Línea 1: Accede al elemento en la posición 0 de la tupla (el primer elemento) y lo imprime.*

*Línea 2: Accede al elemento en la posición 2 de la tupla (el tercer elemento) y lo imprime. Esto demuestra que la lectura de tuplas es igual a la de las listas.*

```python
print('Rojo' in my_tupla)
print(my_tupla.count('Rojo'))
```

*Línea 1: Usa la palabra 'in' para hacerle una pregunta al sistema: "¿Está la palabra 'Rojo' adentro de la tupla?". Devuelve un valor booleano True (Verdadero) o False (Falso).*

*Línea 2: Usa la función .count() para revisar y decirnos cuántas veces exactas se repite la palabra 'Rojo' dentro de esa tupla.*

```python
my_tupla_unitaria = ('Blanco')
print(my_tupla_unitaria)
```

*Línea 1: Intenta crear una tupla de un solo elemento usando paréntesis. (Técnicamente, en Python se necesita poner una coma al final ('Blanco',) para que sea tupla, sino lo tomará como un texto normal).*

*Línea 2: Imprime la variable en pantalla.*

```python
my_tupla = 'Gaspar', 5, 8, 1999
print(my_tupla)
```

*Línea 1: Crea una tupla mezclando texto y números, pero sin usar paréntesis. A esto se le llama "Empaquetado de tupla"; Python agrupa estos valores automáticamente.*

*Línea 2: Imprime la tupla empaquetada (saldrá con los paréntesis puestos automáticamente por el sistema).*

```python
nombre, dia, mes, año = my_tupla
print(nombre)
print(dia)
print(mes)
print(año)
print("Nombre: ", nombre, " - Dia:", dia, " - Mes: ", mes, "- Año: ", año)
```

*Línea 1: Realiza un "Desempaquetado". Toma los 4 elementos guardados en la tupla anterior y los reparte ordenadamente en 4 variables nuevas (nombre, dia, mes, año).*

*Líneas 2 a 5: Imprime cada una de esas variables individuales una debajo de la otra para verificar la asignación.*

*Línea 6: Imprime todas las variables juntas intercaladas con textos en una sola línea ordenada.*

```python
my_lista2 = list(my_tupla)
print(my_lista2)
```

*Línea 1: Usa la función list() para hacer el proceso inverso: tomar una tupla y convertirla en una lista normal que sí pueda ser modificada.*

*Línea 2: Imprime el resultado, el cual volverá a verse encerrado entre corchetes [].*
