```python
names = ['Jenny', 'Alexus', 'Sam', 'Grace']
heights = [61, 70, 67, 64]
```

*Crea dos listas separadas: una con nombres de estudiantes y otra con sus respectivas estaturas en el mismo orden.*

```python
zipStudents = zip(names, heights)
print("zipStudents: ", zipStudents)
```

*Usa la función zip() para emparejar ambas listas. Al imprimirlo, solo se verá un objeto en la memoria de la computadora.*

```python
students = {key:value for key, value in zip(names, heights)}
print(students)
```

*Convierte las listas emparejadas en un diccionario estructurado usando una técnica llamada "Comprensión de diccionarios", y lo muestra.*

```python
drinks = ["espresso", "chai", "decaf", "drip"]
caffeine = [64, 40, 0, 120]
```

*Crea dos nuevas listas independientes para bebidas y su nivel de cafeína.*

```python
zipped_drinks = zip(drinks, caffeine)
print(zipped_drinks)
```

*Empareja las listas de bebidas y cafeína creando un objeto temporal en la memoria.*

```python
drinks_to_caffeine = {key:value for key, value in zipped_drinks}
print(drinks_to_caffeine)
```

*Transforma las bebidas emparejadas en un diccionario real y lo imprime en pantalla.*

```python
songs = ["Like a Rolling Stone", "Satisfaction", "Imagine", "What's Going On", "Respect", "Good Vibrations"]
playcounts = [78, 29, 44, 21, 89, 5]
```

*Crea dos listas: una con títulos de canciones y otra con su número de reproducciones.*

```python
plays = {key:value for key, value in zip(songs, playcounts)}
print(plays)
```

*Empareja las listas y las convierte directamente en un diccionario en un solo paso, mostrándolo después.*
