# Práctica 2: Neurona simple con error cuadrático (neuron1.1.py)

## Objetivo
En este segundo cuaderno conservaremos la misma neurona artificial de la práctica anterior ($y=wx$)[cite: 1]. La diferencia fundamental radica en la estrategia utilizada para aprender el peso[cite: 1]. Pasaremos de una regla directa basada en el error simple, a implementar el error cuadrático y utilizar su derivada para decidir cómo modificar el parámetro[cite: 1].

## Datos de Entrenamiento
Se mantienen los mismos datos de entrenamiento para que la neurona aprenda la relación $t = 2x$[cite: 1]:
* Entradas ($x$): `[1, 2, 3, 4]`[cite: 1]
* Salidas esperadas ($t$): `[2, 4, 6, 8]`[cite: 1]

## Modelo Matemático y Gradientes

El modelo de la neurona se mantiene idéntico:
$$y = wx$$

### Error Cuadrático
En lugar de una simple resta, ahora medimos el desempeño de la neurona utilizando la función de error cuadrático:
$$E = (y - t)^2$$

### Descenso por Gradiente (Regla de la Cadena)
Para saber en qué dirección y con qué magnitud debemos actualizar el peso, necesitamos calcular cómo cambia el error ($E$) cuando cambia el peso ($w$)[cite: 1]. Esto se logra aplicando la regla de la cadena matemática[cite: 1]:

1. Derivada del error respecto a la salida: $\frac{\partial E}{\partial y} = 2(y - t)$[cite: 1]
2. Derivada de la salida respecto al peso: $\frac{\partial y}{\partial w} = x$[cite: 1]
3. Derivada del error respecto al peso: $\frac{\partial E}{\partial w} = \frac{\partial E}{\partial y} \frac{\partial y}{\partial w} = 2(y - t)x$[cite: 1]

Finalmente, el peso se actualiza restando este gradiente multiplicado por la tasa de aprendizaje ($\eta$), lo que matemáticamente se expresa como:
$$w_{nuevo} = w_{anterior} - \eta \frac{\partial E}{\partial w}$$

```python
import time

inputs = [1, 2, 3, 4]
targets = [2, 4, 6, 8]

w = 0.1

epochs = 10

learning_rate = 0.1

pause_time = 1

def predict(i):
    return w * i

for epoch in range(epochs):
    
    predictions = [predict(i) for i in inputs]
    
    errors = [
        (p - t)**2
        for p, t in zip(predictions, targets)
    ]
    
    cost = sum(errors) / len(targets)
    
    errors_d = [
        2 * (p - t)
        for p, t in zip(predictions, targets)
    ]
    
    weight_d = [
        e * i
        for e, i in zip(errors_d, inputs)
    ]
    
    print("\n" + "=" * 65)
    print(f"EPOCA {epoch + 1}")
    print("=" * 65)
    
    for i, t, p, e, ed, wd in zip(inputs, targets, predictions, errors, errors_d, weight_d):
        print(
            f"Entrada: {i:>2} | "
            f"Target: {t:>2} | "
            f"Pred: {p:>8.4f} | "
            f"Error^2: {e:>8.4f} |"
            f"dE/dy: {ed:>8.4f} | "
            f"dE/dw: {wd:>8.4f}"
        )
        
    print("-" * 65)
    print(f"Peso antes de actualizar: {w:.10f}")
    print(f"Costo promedio: {cost:.6f}")
    
    gradient = sum(weight_d) / len(weight_d)
    print(f"Gradiente promedio: {gradient:.6f}")
    
    w -= learning_rate * gradient
    
    print(f"Peso despues de actualizar: {w:.10f}")
    
    time.sleep(pause_time)

test_inputs = [5, 6]
test_targets = [10, 12]

print("\n" + "=" * 65)
print("PRUEBA FINAL")
print("=" * 65)

test_predictions = [
    predict(i)
    for i in test_inputs
]

for i, t, p in zip(test_inputs, test_targets, test_predictions):
    print(
        f"Entrada: {i:>2} | "
        f"Target: {t:>2} | "
        f"Prediccion: {p:>8.4f}"
    )
```

## Análisis de Resultados

En este modelo, observamos un avance conceptual significativo: calcular cómo cambia el error y utilizar esa información matemática para modificar los parámetros[cite: 1]. 

A partir de la ejecución del código, podemos concluir:
1. **Interpretación del gradiente:** El gradiente promedio nos indica la dirección y magnitud para corregir el peso[cite: 1]. Si el gradiente es positivo, el peso disminuye; si es negativo, el peso aumenta, guiando siempre el parámetro hacia la reducción del error[cite: 1].
2. **Convergencia rápida:** A diferencia de la práctica anterior que requirió 30 épocas, al utilizar derivadas del error cuadrático, la neurona logra converger al peso esperado ($w \approx 2.0$) en tan solo 10 épocas.
3. **Generalización exitosa:** Durante la prueba final, el modelo fue capaz de predecir correctamente los valores para $x=5$ y $x=6$, obteniendo salidas de $10.0$ y $12.0$ respectivamente, validando la eficacia del descenso por gradiente.
