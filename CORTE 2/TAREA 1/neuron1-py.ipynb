# Práctica 1: Neurona simple con un solo peso (neuron1.py)

## Objetivo
El propósito de este primer cuaderno es estudiar una neurona artificial muy simple, equipada con una sola entrada y un solo peso, sin utilizar bibliotecas externas. El objetivo es que la neurona aprenda la relación lineal subyacente entre un conjunto de datos de entrenamiento.

## Datos de Entrenamiento
Se utilizarán los siguientes datos:
* Entradas ($x$): `[1, 2, 3, 4]`
* Salidas esperadas ($t$): `[2, 4, 6, 8]`

Al observar los datos, es evidente que la relación ideal a aproximar es $t = 2x$.

## Modelo Matemático y Variables

El modelo de esta neurona se define mediante la ecuación lineal simple:
$$y = wx$$

Donde:
* $x$ es la entrada.
* $w$ es el peso (el parámetro que la neurona debe ajustar).
* $y$ es la predicción o salida calculada.

### Actualización del peso (Aprendizaje)
El peso se inicializa en $w = 0.1$ y se actualizará durante 30 épocas utilizando una regla basada en el error promedio. 
El error individual se calcula como la diferencia entre el valor esperado ($t$) y la predicción ($p$):
$$e = t - p$$

La regla de actualización para el parámetro es:
$$w_{nuevo} = w_{anterior} + \eta \cdot cost$$
Donde $\eta$ (eta) representa la tasa de aprendizaje (`learning_rate = 0.1`) y $cost$ es el error promedio de la época actual.

```python
import time

inputs = [1, 2, 3, 4]
targets = [2, 4, 6, 8]

w = 0.1

learning_rate = 0.1

epochs = 30

pause_time = 1

def predict(i):
    return w * i

for epoch in range(epochs):
    
    predictions = [predict(i) for i in inputs]
    
    errors = [
        t - p
        for p, t in zip(predictions, targets)
    ]
    
    cost = sum(errors) / len(targets)
    
    print("\n" + "=" * 60)
    print(f"EPOCA {epoch + 1}")
    print("=" * 60)
    
    for i, t, p, e in zip(inputs, targets, predictions, errors):
        print(
            f"Entrada: {i:>2} | "
            f"Target: {t:>2} | "
            f"Prediccion: {p:>8.4f} | "
            f"Error: {e:>8.4f}"
        )
        
    print("-" * 60)
    print(f"Peso antes de actualizar: {w:.10f}")
    print(f"Error promedio: {cost:.6f}")
    
    w += learning_rate * cost
    
    print(f"Peso despues de actualizar: {w:.10f}")
    
    time.sleep(pause_time)

test_inputs = [5, 6]
test_targets = [10, 12]

print("\n" + "=" * 60)
print("PRUEBA FINAL")
print("=" * 60)

test_predictions = [
    predict(i)
    for i in test_inputs
]

for i, t, p in zip(test_inputs, test_targets, test_predictions):
    print(
        f"Entrada: {i:>2} | "
        f"Target: {t:>2} | "
        f"Prediccion: {p:>8.4f}"
    )
```

## Análisis de Resultados

Al ejecutar el modelo y finalizar las 30 épocas de entrenamiento, podemos observar lo siguiente:
1. **Evolución del peso:** El parámetro $w$ inició con un valor de `0.1` (muy lejano al objetivo) y progresivamente aumentó iteración tras iteración hasta aproximarse al valor de `2.0`.
2. **Reducción del error:** A medida que el peso se ajustaba, el error promedio fue disminuyendo gradualmente hasta acercarse a cero, demostrando un aprendizaje estable.
3. **Generalización (Prueba Final):** Al ingresar datos nuevos que la red no había visto durante el entrenamiento ($x = 5$ y $x = 6$), la neurona logró predecir valores extremadamente precisos con respecto a las salidas esperadas ($10$ y $12$). Esto confirma que la neurona extrajo exitosamente la relación matemática subyacente.
