# Práctica 3: Neurona simple con peso y sesgo (neuron1.2.py)

## Objetivo
En esta práctica se amplía el modelo utilizado anteriormente[cite: 26]. Ahora se incorpora un segundo parámetro denominado sesgo o bias, representado mediante $b$[cite: 26]. La diferencia fundamental respecto a las prácticas anteriores es que ahora el proceso de entrenamiento debe ajustar simultáneamente dos parámetros[cite: 26].

## Datos de Entrenamiento
Los datos utilizados para el entrenamiento son $x = [1, 2, 3, 4]$ y $t = [4, 6, 8, 10]$[cite: 26]. 
Al observar estos datos puede identificarse la relación $t = 2x + 2$[cite: 26]. Por tanto, la neurona debe aprender aproximadamente los parámetros $w \approx 2$ y $b \approx 2$[cite: 26].

## Modelo Matemático y Actualización de Parámetros

El nuevo modelo es $y = wx + b$[cite: 26]. El parámetro $b$ permite desplazar la relación obtenida[cite: 29].

### Derivadas y Gradientes
Para ajustar ambos parámetros simultáneamente, calculamos sus derivadas parciales. 
Para el peso, mantenemos el cálculo de la práctica anterior. Para el sesgo, dado que $\frac{\partial y}{\partial b} = 1$, al aplicar la regla de la cadena la derivada del error respecto al sesgo resulta en[cite: 32]:
$$\frac{\partial E}{\partial b} = 2(y - t)$$

### Actualización Simultánea
El peso se actualiza mediante $w_{nuevo} = w_{anterior} - \eta \frac{\partial E}{\partial w}$, mientras que el sesgo se actualiza mediante $b_{nuevo} = b_{anterior} - \eta \frac{\partial E}{\partial b}$[cite: 33]. Ambos parámetros se modifican en dirección opuesta a su gradiente para minimizar la función de costo.

```python
import time

inputs = [1, 2, 3, 4]
targets = [4, 6, 8, 10]

w = 0.1
b = 0.3

epochs = 300
learning_rate = 0.1
pause_time = 0.1

def predict(i):
    return w * i + b

for epoch in range(epochs):
    predictions = [
        predict(i)
        for i in inputs
    ]
    
    errors = [
        (p - t)**2
        for p, t in zip(predictions, targets)
    ]
    
    cost = sum(errors) / len(targets)
    
    errors_d = [
        2 * (p - t)
        for p, t in zip(predictions, targets)
    ]
    
    weight_d = [
        e * i
        for e, i in zip(errors_d, inputs)
    ]
    
    bias_d = [
        e
        for e in errors_d
    ]
    
    print("\n" + "=" * 70)
    print(f"EPOCA {epoch + 1}")
    print("=" * 70)
    
    for i, t, p, e, wd, bd in zip(inputs, targets, predictions, errors, weight_d, bias_d):
        print(
            f"Entrada: {i:>2} | "
            f"Target: {t:>2} | "
            f"Pred: {p:>8.4f} | "
            f"Error^2: {e:>8.4f} |"
            f"dE/dw: {wd:>9.4f} | "
            f"dE/db: {bd:>9.4f}"
        )
        
    gradient_w = sum(weight_d) / len(weight_d)
    gradient_b = sum(bias_d) / len(bias_d)
    
    print("-" * 70)
    print(f"Peso antes de actualizar: {w:.10f}")
    print(f"Bias antes de actualizar: {b:.10f}")
    print(f"Costo promedio: {cost:.6f}")
    print(f"Gradiente de w: {gradient_w:.6f}")
    print(f"Gradiente de b: {gradient_b:.6f}")
    
    w -= learning_rate * gradient_w
    b -= learning_rate * gradient_b
    
    print(f"Peso despues de actualizar: {w:.10f}")
    print(f"Bias despues de actualizar: {b:.10f}")
    
    time.sleep(pause_time)

test_inputs = [5, 6]

print("\n" + "=" * 70)
print("PRUEBA FINAL")
print("=" * 70)

test_predictions = [
    predict(i)
    for i in test_inputs
]

for i, p in zip(test_inputs, test_predictions):
    print(
        f"Entrada: {i:>2} | "
        f"Prediccion: {p:>8.4f}"
    )
```

## Análisis de Resultados

Tras someter el modelo a 300 épocas de entrenamiento, podemos destacar:
1. **Convergencia Dual:** Se logró un ajuste exitoso de ambos parámetros de forma paralela. Iniciando con valores lejanos ($w=0.1$ y $b=0.3$), la neurona logró converger a los valores meta de $w \approx 2$ y $b \approx 2$. 
2. **Importancia del Sesgo:** Al integrar el sesgo ($b$), la red adquirió la capacidad de desplazar la curva, permitiéndole mapear relaciones que no pasan por el origen (0,0). El modelo final aproximó de forma excelente la ecuación subyacente $y = 2x + 2$.
3. **Validación:** En la prueba final con $x=5$ y $x=6$, la red infirió automáticamente los valores esperados de $12.0$ y $14.0$, confirmando su capacidad de generalización.
